# Imports + Cesty

```python
import sys
from pathlib import Path

sys.path.insert(0, "scripts")
from octa_mae_eval import run_encoder_eval_final

ENCODER_ROOT = Path().resolve()
RESULTS      = ENCODER_ROOT / "results"

EXCEL_PATH = ENCODER_ROOT / "data" / "master_excels" / "master_table.xlsx"
DATA_ROOT  = ENCODER_ROOT / "data"

P1_BASELINE = RESULTS / "phase1" / "baseline" / "pretrained" / "final_encoder.pth"
P1_BALANCED = RESULTS / "phase1" / "balanced" / "pretrained" / "final_encoder.pth"
P2_BASELINE = RESULTS / "phase2" / "baseline" / "pretrained" / "final_encoder.pth"
P2_BALANCED = RESULTS / "phase2" / "balanced" / "pretrained" / "final_encoder.pth"
P2_GRL      = RESULTS / "phase2" / "grl"      / "pretrained" / "final_encoder.pth"

paths = {
    "P1 Baseline": P1_BASELINE,
    "P1 Balanced": P1_BALANCED,
    "P2 Baseline": P2_BASELINE,
    "P2 Balanced": P2_BALANCED,
    "P2 GRL"     : P2_GRL,
}

print(f"Encoder root: {ENCODER_ROOT}\n")
for name, path in paths.items():
    status = "OK" if path.exists() else "NENAJDENY"
    print(f"  {name:<15} {status:<12} {path}")
```

## Bunka 7 — Porovnanie vsetkych experimentov

```python
experiments = {
    "Exp1 P1 Baseline"         : out_exp1,
    "Exp2 P1 Balanced"         : out_exp2,
    "Exp3 P1Base -> P2Base"    : out_exp3,
    "Exp4 P1Bal  -> P2Bal"     : out_exp4,
    "Exp5 P1Bal  -> P2GRL"     : out_exp5,
}

metrics = [
    "SVP only - Disease",
    "SVP only - Dataset",
    "DCP only - Disease",
    "DCP only - Dataset",
    "Concat - Disease",
    "Concat - Dataset",
    "Single - SVP vs DCP",
]

col_w = 28
header = f"{'Metrika':<{col_w}}" + "".join(f"{name:>22}" for name in experiments)
print(header)
print("-" * len(header))

for metric in metrics:
    row = f"{metric:<{col_w}}"
    for exp_out in experiments.values():
        val = exp_out["results"].get(metric, {}).get("bal_acc", None)
        row += f"{val:>22.4f}" if val is not None else f"{'N/A':>22}"
    print(row)
```
